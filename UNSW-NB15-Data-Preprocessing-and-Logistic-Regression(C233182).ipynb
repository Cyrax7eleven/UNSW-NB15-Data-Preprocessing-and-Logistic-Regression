```python
# Select numerical features only
numeric_cols = train_clean.select_dtypes(
    include=["int64", "float64"]
).columns.drop("label")

outlier_results = []

for col in numeric_cols:

    Q1 = train_clean[col].quantile(0.25)
    Q3 = train_clean[col].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outlier_count = (
        (train_clean[col] < lower_bound) |
        (train_clean[col] > upper_bound)
    ).sum()

    outlier_percentage = (
        outlier_count / len(train_clean) * 100
    )

    outlier_results.append([
        col,
        outlier_count,
        round(outlier_percentage, 2)
    ])

outlier_df = pd.DataFrame(
    outlier_results,
    columns=["Feature", "Outlier Count", "Outlier Percentage"]
)

outlier_df = outlier_df.sort_values(
    by="Outlier Percentage",
    ascending=False
)

print(outlier_df)
```

```python
# Features suitable for log transformation

log_features = [
    'trans_depth',
    'response_body_len',
    'sbytes',
    'sloss',
    'dloss',
    'spkts',
    'dbytes',
    'dpkts',
    'dinpkt',
    'djit',
    'ct_flw_http_mthd',
    'sjit',
    'sload',
    'sinpkt',
    'dur',
    'synack',
    'ackdat',
    'dload',
    'tcprtt',
    'smean',
    'rate',
    'dmean'
]

print("Features selected for log transformation:")
print(log_features)

print("\nTotal features:", len(log_features))
```

```python
# Make copies so original cleaned data remains unchanged

train_transformed = train_clean.copy()
test_transformed = test_clean.copy()

# Apply log(1 + x) transformation
train_transformed[log_features] = np.log1p(
    train_transformed[log_features]
)

test_transformed[log_features] = np.log1p(
    test_transformed[log_features]
)

print("Log transformation applied successfully!")

print("\nTraining shape:", train_transformed.shape)
print("Testing shape:", test_transformed.shape)
```

```python
# Skewness before transformation
before_skew = train_clean[log_features].skew()

# Skewness after transformation
after_skew = train_transformed[log_features].skew()

# Create comparison table
skew_comparison = pd.DataFrame({
    'Feature': log_features,
    'Before Skewness': before_skew.values,
    'After Skewness': after_skew.values
})

# Absolute values for easier comparison
skew_comparison['Abs Before'] = skew_comparison['Before Skewness'].abs()
skew_comparison['Abs After'] = skew_comparison['After Skewness'].abs()

# Improvement percentage
skew_comparison['Reduction %'] = (
    (skew_comparison['Abs Before'] - skew_comparison['Abs After'])
    / skew_comparison['Abs Before']
) * 100

# Sort by highest original skewness
skew_comparison = skew_comparison.sort_values(
    by='Abs Before',
    ascending=False
)

print(skew_comparison)
```

```python
# Find categorical columns

categorical_cols = train_transformed.select_dtypes(
    include=['object']
).columns.tolist()

print("Categorical columns:")
print(categorical_cols)

print("\nUnique values in each categorical column:")

for col in categorical_cols:
    print(f"{col}: {train_transformed[col].nunique()} unique values")
```

```python
from sklearn.preprocessing import OneHotEncoder

# Create encoder
encoder = OneHotEncoder(
    handle_unknown='ignore',
    sparse_output=False
)

# Fit on training categorical data
encoder.fit(train_transformed[categorical_cols])

# Transform training and testing data
train_encoded = encoder.transform(
    train_transformed[categorical_cols]
)

test_encoded = encoder.transform(
    test_transformed[categorical_cols]
)

print("Training encoded shape:", train_encoded.shape)
print("Testing encoded shape:", test_encoded.shape)

print("\nTotal encoded features:")
print(len(encoder.get_feature_names_out(categorical_cols)))
```

```python
import pandas as pd

# Select numerical columns
numeric_cols = train_transformed.select_dtypes(
    include=['int64', 'float64']
).columns.drop('label')

print("Number of numerical features:", len(numeric_cols))

# Create DataFrame from encoded features
encoded_feature_names = encoder.get_feature_names_out(categorical_cols)

train_encoded_df = pd.DataFrame(
    train_encoded,
    columns=encoded_feature_names,
    index=train_transformed.index
)

test_encoded_df = pd.DataFrame(
    test_encoded,
    columns=encoded_feature_names,
    index=test_transformed.index
)

# Combine numerical + encoded categorical features
X_train = pd.concat(
    [
        train_transformed[numeric_cols],
        train_encoded_df
    ],
    axis=1
)

X_test = pd.concat(
    [
        test_transformed[numeric_cols],
        test_encoded_df
    ],
    axis=1
)

# Target variable
y_train = train_transformed['label']
y_test = test_transformed['label']

print("\nFINAL DATASET SHAPE:")
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)

print("\nTarget shape:")
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.preprocessing import StandardScaler

# Create scaler
scaler = StandardScaler()

# Fit scaler ONLY on training numerical features
X_train_scaled_numeric = scaler.fit_transform(
    X_train[numeric_cols]
)

# Transform testing numerical features
X_test_scaled_numeric = scaler.transform(
    X_test[numeric_cols]
)

print("Scaling completed!")

print("Training numeric shape:", X_train_scaled_numeric.shape)
print("Testing numeric shape:", X_test_scaled_numeric.shape)
```

```python
# Convert encoded categorical data to DataFrame

train_encoded_df = pd.DataFrame(
    train_encoded,
    columns=encoder.get_feature_names_out(categorical_cols),
    index=train_transformed.index
)

test_encoded_df = pd.DataFrame(
    test_encoded,
    columns=encoder.get_feature_names_out(categorical_cols),
    index=test_transformed.index
)


# Combine scaled numerical features + encoded categorical features

X_train_final = pd.concat(
    [
        pd.DataFrame(
            X_train_scaled_numeric,
            columns=numeric_cols,
            index=train_transformed.index
        ),
        train_encoded_df
    ],
    axis=1
)

X_test_final = pd.concat(
    [
        pd.DataFrame(
            X_test_scaled_numeric,
            columns=numeric_cols,
            index=test_transformed.index
        ),
        test_encoded_df
    ],
    axis=1
)


# Target

y_train = train_transformed["label"]
y_test = test_transformed["label"]


print("FINAL TRAIN SHAPE:", X_train_final.shape)
print("FINAL TEST SHAPE:", X_test_final.shape)

print("\nTarget:")
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

# Create Logistic Regression model
model_lr = LogisticRegression(
    max_iter=1000,
    random_state=42
)

# Train the model
model_lr.fit(X_train_final, y_train)

print("Model training completed!")
```

```python
# Make predictions on test data
y_pred = model_lr.predict(X_test_final)

print("Prediction completed!")
```

```python
# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

print("Logistic Regression Accuracy:", accuracy)
```
